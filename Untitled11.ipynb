```python
@dataclass
class ControlAxis:
  index: int
  label: str
  vector: np.ndarray

  positive_example: str
  negative_example: str

  layer_idx: int
  steering_layers: List[int]
  layer_weights: List[float]

  # PCA metadata
  variance: float = 0.0
  scale: float = 1.0
  source: str = "pca"
  coherence: float = 0.0

  # Sample extremes before steering, useful for debugging/paper logs
  positive_sample: str = ""
  negative_sample: str = ""


class PCAControlSystem:
  """
  Paper-style PCA activation steering:

  1. For a single prompt, sample stochastic completions.
  2. Extract penultimate-layer hidden activations over generated tokens.
  3. Mean-pool each completion into one activation vector.
  4. Run PCA over the activation matrix.
  5. Keep top PCs as activation steering directions.
  6. Generate +/- pole outputs.
  7. Ask local LLM judge to label and orient each axis.
  """

  def __init__(
      self,
      model_id: str = "Qwen/Qwen2.5-3B-Instruct",
      labeler_id: Optional[str] = None,
      target_layer: Optional[int] = None,
      device: str = device,
      use_4bit_labeler: bool = True,
  ):
      self.model_id = model_id
      self.labeler_id = labeler_id
      self.device = device

      print(f"\nLoading engine model: {model_id}")

      self.tokenizer = AutoTokenizer.from_pretrained(
          model_id,
          trust_remote_code=True,
      )
      if self.tokenizer.pad_token is None:
          self.tokenizer.pad_token = self.tokenizer.eos_token

      engine_kwargs = {
          "trust_remote_code": True,
      }

      if self.device == "cuda":
          engine_kwargs.update({
              "torch_dtype": torch.float16,
              "device_map": "auto",
          })
      else:
          engine_kwargs.update({
              "torch_dtype": torch.float32,
          })

      self.model = AutoModelForCausalLM.from_pretrained(
          model_id,
          **engine_kwargs,
      )
      self.model.eval()

      self.layers = self._get_layers(self.model)
      self.embed_tokens = self._get_embed_tokens(self.model)

      self.penultimate_layer_idx = len(self.layers) - 2
      self.target_layer = target_layer if target_layer is not None else self.penultimate_layer_idx

      print(f"Total transformer layers: {len(self.layers)}")
      print(f"Using target / penultimate layer index: {self.target_layer}")

      # Labeler model. If labeler_id is None, reuse engine model.
      if labeler_id is None or labeler_id == model_id:
          print("\nUsing engine model as judge/labeler.")
          self.labeler_tokenizer = self.tokenizer
          self.labeler_model = self.model
      else:
          print(f"\nLoading separate judge/labeler model: {labeler_id}")

          self.labeler_tokenizer = AutoTokenizer.from_pretrained(
              labeler_id,
              trust_remote_code=True,
          )
          if self.labeler_tokenizer.pad_token is None:
              self.labeler_tokenizer.pad_token = self.labeler_tokenizer.eos_token

          labeler_kwargs = {
              "trust_remote_code": True,
          }

          if self.device == "cuda":
              if use_4bit_labeler:
                  try:
                      from transformers import BitsAndBytesConfig

                      quantization_config = BitsAndBytesConfig(
                          load_in_4bit=True,
                          bnb_4bit_compute_dtype=torch.float16,
                      )
                      labeler_kwargs.update({
                          "device_map": "auto",
                          "quantization_config": quantization_config,
                      })
                      print("Using 4-bit quantization for labeler.")
                  except Exception as e:
                      print("bitsandbytes unavailable; loading labeler in fp16 instead.")
                      print("Reason:", e)
                      labeler_kwargs.update({
                          "device_map": "auto",
                          "torch_dtype": torch.float16,
                      })
              else:
                  labeler_kwargs.update({
                      "device_map": "auto",
                      "torch_dtype": torch.float16,
                  })
          else:
              labeler_kwargs.update({
                  "torch_dtype": torch.float32,
              })

          self.labeler_model = AutoModelForCausalLM.from_pretrained(
              labeler_id,
              **labeler_kwargs,
          )
          self.labeler_model.eval()

      print("\nPCAControlSystem ready.")

  # ----------------------------
  # Model structure helpers
  # ----------------------------
  def _get_layers(self, model):
      candidates = [
          ("model", "layers"),
          ("transformer", "h"),
          ("gpt_neox", "layers"),
      ]

      for path in candidates:
          obj = model
          ok = True
          for attr in path:
              if not hasattr(obj, attr):
                  ok = False
                  break
              obj = getattr(obj, attr)
          if ok:
              return obj

      raise ValueError("Could not find transformer layers for this model architecture.")

  def _get_embed_tokens(self, model):
      candidates = [
          ("model", "embed_tokens"),
          ("transformer", "wte"),
          ("gpt_neox", "embed_in"),
      ]

      for path in candidates:
          obj = model
          ok = True
          for attr in path:
              if not hasattr(obj, attr):
                  ok = False
                  break
              obj = getattr(obj, attr)
          if ok:
              return obj

      raise ValueError("Could not find embedding module for this model architecture.")

  def _model_input_device(self, model):
      return next(model.parameters()).device

  # ----------------------------
  # Text / token helpers
  # ----------------------------
  def _chat_text(self, tokenizer, messages):
      try:
          return tokenizer.apply_chat_template(
              messages,
              tokenize=False,
              add_generation_prompt=True,
          )
      except Exception:
          # Fallback for models without chat templates.
          rendered = []
          for m in messages:
              rendered.append(f"{m['role'].upper()}: {m['content']}")
          rendered.append("ASSISTANT:")
          return "\n".join(rendered)

  def _prepare_inputs(self, tokenizer, text, model):
      inputs = tokenizer(text, return_tensors="pt")
      return inputs.to(self._model_input_device(model))

  def _decode_new_tokens(self, tokenizer, output_ids, prompt_len):
      generated_ids = output_ids[0][prompt_len:]
      text = tokenizer.decode(generated_ids, skip_special_tokens=True)
      return self._clean_output(text)

  def _clean_output(self, text: str) -> str:
      text = text.strip()

      patterns = [
          r"^Sure[,\!]?\s*",
          r"^Certainly[,\!]?\s*",
          r"^Here(?:'s| is)\s+(?:a|the)?\s*.*?:\s*",
          r"^Response:\s*",
          r"^Output:\s*",
          r"^Assistant:\s*",
      ]

      for pattern in patterns:
          text = re.sub(pattern, "", text, flags=re.IGNORECASE).strip()

      if text.startswith('"') and text.endswith('"'):
          text = text[1:-1].strip()
      elif text.endswith('"'):
          text = text[:-1].strip()

      return text

  def _clean_label(self, label: str) -> str:
      label = str(label).strip()
      words = re.findall(r"[A-Za-z][A-Za-z\-]*", label)

      banned = {
          "the", "a", "an", "style", "axis", "label", "difference",
          "text", "response", "more", "less", "positive", "negative",
          "pole", "assistant", "json", "attribute",
      }

      words = [w for w in words if w.lower() not in banned]

      if not words:
          return "Style"

      # Prefer short UI-friendly labels.
      return " ".join(w.capitalize() for w in words[:2])

  # ----------------------------
  # Plain generation
  # ----------------------------
  def generate_text(
      self,
      prompt: str,
      system_prompt: str = "You are a helpful assistant.",
      max_new_tokens: int = 180,
      do_sample: bool = True,
      temperature: float = 0.4,
      top_p: float = 0.9,
      repetition_penalty: float = 1.08,
  ) -> str:
      messages = [
          {"role": "system", "content": system_prompt},
          {"role": "user", "content": prompt},
      ]

      input_text = self._chat_text(self.tokenizer, messages)
      inputs = self._prepare_inputs(self.tokenizer, input_text, self.model)
      prompt_len = inputs["input_ids"].shape[1]

      gen_kwargs = {
          "max_new_tokens": max_new_tokens,
          "do_sample": do_sample,
          "repetition_penalty": repetition_penalty,
          "pad_token_id": self.tokenizer.eos_token_id,
      }

      if do_sample:
          gen_kwargs.update({
              "temperature": temperature,
              "top_p": top_p,
          })

      with torch.no_grad():
          outputs = self.model.generate(
              **inputs,
              **gen_kwargs,
          )

      if hasattr(outputs, "sequences"):
          sequences = outputs.sequences
      else:
          sequences = outputs

      return self._decode_new_tokens(self.tokenizer, sequences, prompt_len)

  # ----------------------------
  # Generation + activation extraction
  # ----------------------------
  def generate_with_activation(
      self,
      prompt: str,
      system_prompt: str = "You are a helpful assistant.",
      layer_idx: Optional[int] = None,
      max_new_tokens: int = 180,
      temperature: float = 1.15,
      top_p: float = 0.95,
      repetition_penalty: float = 1.05,
  ) -> Tuple[str, np.ndarray]:
      """
      Generate one stochastic completion and extract the target-layer
      hidden activations over the generated-token window.

      Important indexing detail:
      Hugging Face hidden_states usually contains:
        hidden_states[0] = embedding output
        hidden_states[k + 1] = output after transformer block k

      So if we steer transformer block layer_idx,
      we extract hidden_states[layer_idx + 1].
      """

      if layer_idx is None:
          layer_idx = self.target_layer

      messages = [
          {"role": "system", "content": system_prompt},
          {"role": "user", "content": prompt},
      ]

      input_text = self._chat_text(self.tokenizer, messages)
      inputs = self._prepare_inputs(self.tokenizer, input_text, self.model)
      prompt_len = inputs["input_ids"].shape[1]

      with torch.no_grad():
          outputs = self.model.generate(
              **inputs,
              max_new_tokens=max_new_tokens,
              do_sample=True,
              temperature=temperature,
              top_p=top_p,
              repetition_penalty=repetition_penalty,
              pad_token_id=self.tokenizer.eos_token_id,
          )

      if hasattr(outputs, "sequences"):
          sequence_tensor = outputs.sequences
      else:
          sequence_tensor = outputs

      generated_text = self._decode_new_tokens(
          self.tokenizer,
          sequence_tensor,
          prompt_len,
      )

      # Forward pass over the full prompt+completion to get hidden states.
      sequence_tensor = sequence_tensor.to(self._model_input_device(self.model))

      with torch.no_grad():
          forward_out = self.model(
              input_ids=sequence_tensor,
              output_hidden_states=True,
              return_dict=True,
          )

      hidden_states = forward_out.hidden_states
      hidden_state_idx = layer_idx + 1

      if hidden_states is None or hidden_state_idx >= len(hidden_states):
          raise RuntimeError(
              f"Could not extract hidden_states[{hidden_state_idx}]. "
              f"Available hidden states: {0 if hidden_states is None else len(hidden_states)}"
          )

      target_hidden = hidden_states[hidden_state_idx]  # [1, total_seq_len, hidden_dim]

      if sequence_tensor.shape[1] > prompt_len:
          generated_hidden = target_hidden[0, prompt_len:, :]
      else:
          generated_hidden = target_hidden[0, -1:, :]

      # Paper methodology: sequence-average pool over generated-token window.
      pooled = generated_hidden.mean(dim=0).detach().float().cpu().numpy()

      return generated_text, pooled

  def sample_completion_cloud(
      self,
      prompt: str,
      num_samples: int = 30,
      layer_idx: Optional[int] = None,
      temperature: float = 1.15,
      top_p: float = 0.95,
      max_new_tokens: int = 180,
  ) -> Tuple[List[str], np.ndarray]:
      """
      Sample stochastic completions and return:
        texts: List[str]
        activations: np.ndarray shape [num_samples, hidden_dim]
      """

      if layer_idx is None:
          layer_idx = self.target_layer

      texts = []
      activations = []
      seen = set()

      attempts = 0
      max_attempts = num_samples * 3

      print(f"\nSampling {num_samples} stochastic completions at temperature={temperature}...")

      while len(texts) < num_samples and attempts < max_attempts:
          attempts += 1

          text, vec = self.generate_with_activation(
              prompt=prompt,
              system_prompt="You are a helpful assistant. Answer naturally.",
              layer_idx=layer_idx,
              max_new_tokens=max_new_tokens,
              temperature=temperature,
              top_p=top_p,
          )

          normalized = re.sub(r"\s+", " ", text.strip().lower())

          # Skip empty or duplicate-ish completions.
          if not normalized or normalized in seen:
              continue

          seen.add(normalized)
          texts.append(text)
          activations.append(vec)

          print(f"[{len(texts):02d}/{num_samples}] {text[:120].replace(chr(10), ' ')}...")

      if len(texts) < 4:
          raise ValueError(
              f"Only got {len(texts)} usable completions; need at least 4 for PCA."
          )

      return texts, np.stack(activations, axis=0)

  # ----------------------------
  # Steering
  # ----------------------------
  def build_steering_layers(
      self,
      center_layer: int,
      radius: int = 0,
      decay: float = 0.7,
  ) -> Tuple[List[int], List[float]]:
      """
      Paper-faithful default is radius=0: steer only penultimate layer.
      You can experiment with radius=1 or radius=2 later.
      """

      total_layers = len(self.layers)
      steering_layers = []
      weights = []

      for offset in range(-radius, radius + 1):
          layer = center_layer + offset
          if layer < 0 or layer >= total_layers:
              continue

          steering_layers.append(layer)
          weights.append(decay ** abs(offset))

      weights = np.array(weights, dtype=np.float32)
      weights = weights / weights.sum()

      return steering_layers, weights.tolist()

  def clear_forward_hooks(self):
      # Notebook safety: remove stale hooks from previous interrupted runs.
      for layer in self.layers:
          layer._forward_hooks.clear()

  def steer_generation(
      self,
      prompt: str,
      axis: ControlAxis,
      coefficient: float = 0.0,
      max_new_tokens: int = 180,
      system_prompt: str = "You are a helpful assistant.",
      temperature: float = 0.4,
      top_p: float = 0.9,
      repetition_penalty: float = 1.08,
  ) -> str:
      """
      coefficient is in PCA standard-deviation units.

      For PCA axes:
        delta = coefficient * sqrt(explained_variance) * unit_pc_vector

      The frontend can still use a -50..50 slider; the API cell maps that
      UI value into about -3..3 coefficient units.
      """

      if abs(float(coefficient)) < 1e-9:
          return self.generate_text(
              prompt=prompt,
              system_prompt=system_prompt,
              max_new_tokens=max_new_tokens,
              do_sample=True,
              temperature=temperature,
              top_p=top_p,
              repetition_penalty=repetition_penalty,
          )

      self.clear_forward_hooks()

      input_device = self._model_input_device(self.model)
      dtype = torch.float16 if self.device == "cuda" else torch.float32

      base_vector = torch.tensor(
          axis.vector,
          dtype=dtype,
          device=input_device,
      )

      # PCA scale is sqrt(explained_variance).
      # For custom non-PCA axes, scale may be 1.
      base_vector = base_vector * float(axis.scale) * float(coefficient)

      handles = []

      for layer_idx, layer_weight in zip(axis.steering_layers, axis.layer_weights):
          layer_delta = base_vector * float(layer_weight)

          def make_hook(delta):
              def steering_hook(module, module_input, module_output):
                  if isinstance(module_output, tuple):
                      hidden_states = module_output[0]
                      rest = module_output[1:]
                  else:
                      hidden_states = module_output
                      rest = None

                  hidden_states = hidden_states.clone()

                  # Add to the final position in the current forward pass.
                  # During generation this affects the next-token distribution.
                  delta_local = delta.to(
                      device=hidden_states.device,
                      dtype=hidden_states.dtype,
                  )
                  hidden_states[:, -1, :] = hidden_states[:, -1, :] + delta_local

                  if rest is None:
                      return hidden_states
                  return (hidden_states,) + rest

              return steering_hook

          handle = self.layers[layer_idx].register_forward_hook(make_hook(layer_delta))
          handles.append(handle)

      try:
          messages = [
              {"role": "system", "content": system_prompt},
              {"role": "user", "content": prompt},
          ]

          input_text = self._chat_text(self.tokenizer, messages)
          inputs = self._prepare_inputs(self.tokenizer, input_text, self.model)
          prompt_len = inputs["input_ids"].shape[1]

          with torch.no_grad():
              outputs = self.model.generate(
                  **inputs,
                  max_new_tokens=max_new_tokens,
                  do_sample=True,
                  temperature=temperature,
                  top_p=top_p,
                  repetition_penalty=repetition_penalty,
                  pad_token_id=self.tokenizer.eos_token_id,
              )

          if hasattr(outputs, "sequences"):
              sequences = outputs.sequences
          else:
              sequences = outputs

          generated = self._decode_new_tokens(self.tokenizer, sequences, prompt_len)

          if not generated:
              generated = "[Empty generation. Try a smaller coefficient.]"

          return generated

      finally:
          for h in handles:
              h.remove()

  # ----------------------------
  # Judge / labeler
  # ----------------------------
  def labeler_generate(
      self,
      prompt: str,
      system_prompt: str = "You are a concise expert linguistic judge.",
      max_new_tokens: int = 180,
  ) -> str:
      messages = [
          {"role": "system", "content": system_prompt},
          {"role": "user", "content": prompt},
      ]

      input_text = self._chat_text(self.labeler_tokenizer, messages)
      inputs = self._prepare_inputs(self.labeler_tokenizer, input_text, self.labeler_model)
      prompt_len = inputs["input_ids"].shape[1]

      with torch.no_grad():
          outputs = self.labeler_model.generate(
              **inputs,
              max_new_tokens=max_new_tokens,
              do_sample=False,
              pad_token_id=self.labeler_tokenizer.eos_token_id,
          )

      if hasattr(outputs, "sequences"):
          sequences = outputs.sequences
      else:
          sequences = outputs

      return self._decode_new_tokens(self.labeler_tokenizer, sequences, prompt_len)

  def parse_json_object(self, text: str) -> Optional[Dict[str, Any]]:
      try:
          start = text.find("{")
          end = text.rfind("}")
          if start == -1 or end == -1 or end <= start:
              return None
          return json.loads(text[start:end + 1])
      except Exception:
          return None

  def label_and_orient_axis(
      self,
      user_prompt: str,
      negative_text: str,
      baseline_text: str,
      positive_text: str,
  ) -> Dict[str, Any]:
      """
      Ask the local LLM judge to:
        - label the stylistic axis
        - decide whether the + pole has more of that label
        - rate coherence
      """

      judge_prompt = f"""
You are labeling an activation-steering axis for a language model.

The same user prompt was answered three times.

USER PROMPT:
{user_prompt}

NEGATIVE POLE, generated with coefficient -c:
\"\"\"
{negative_text}
\"\"\"

BASELINE, generated with coefficient 0:
\"\"\"
{baseline_text}
\"\"\"

POSITIVE POLE, generated with coefficient +c:
\"\"\"
{positive_text}
\"\"\"

Task:
Identify the SINGLE most salient stylistic or rhetorical attribute that changes between the negative and positive poles.

Important rules:
- Do NOT name the topic or subject matter.
- Prefer labels like Formality, Detail, Vividness, Concision, Enthusiasm, Politeness, Directness, Complexity, Specificity, Warmth, Objectivity.
- The label should be 1 or 2 words.
- Decide whether the POSITIVE POLE has MORE of that label than the NEGATIVE POLE.
- Rate axis coherence from 0 to 10.

Return ONLY valid JSON with exactly this schema:
{{
"label": "one or two words",
"positive_pole_has_more_label": true,
"coherence": 0,
"rationale": "one short sentence"
}}
""".strip()

      raw = self.labeler_generate(
          prompt=judge_prompt,
          system_prompt="You are a strict JSON-only linguistic evaluator.",
          max_new_tokens=180,
      )

      data = self.parse_json_object(raw)

      if data is None:
          print("\n[Judge parse warning] Raw judge output:")
          print(raw)

          return {
              "label": "Style",
              "positive_pole_has_more_label": True,
              "coherence": 3.0,
              "rationale": "Judge output could not be parsed.",
              "raw": raw,
          }

      label = self._clean_label(data.get("label", "Style"))

      pos_has_more = data.get("positive_pole_has_more_label", True)
      if isinstance(pos_has_more, str):
          pos_has_more = pos_has_more.strip().lower() in {"true", "yes", "positive", "+", "1"}

      try:
          coherence = float(data.get("coherence", 5.0))
      except Exception:
          coherence = 5.0

      coherence = max(0.0, min(10.0, coherence))

      return {
          "label": label,
          "positive_pole_has_more_label": bool(pos_has_more),
          "coherence": coherence,
          "rationale": str(data.get("rationale", "")),
          "raw": raw,
      }

  # ----------------------------
  # PCA discovery
  # ----------------------------
  def discover_pca_axes(
      self,
      prompt: str,
      num_samples: int = 30,
      n_axes: int = 2,
      pole_coefficient: float = 2.0,
      layer_idx: Optional[int] = None,
      sample_temperature: float = 1.15,
      sample_top_p: float = 0.95,
      max_new_tokens: int = 180,
      steering_radius: int = 0,
  ) -> Tuple[List[ControlAxis], List[str], str]:
      """
      Main paper-method function.

      Returns:
        axes, sampled_variations, baseline_text
      """

      if layer_idx is None:
          layer_idx = self.target_layer

      n_axes = min(int(n_axes), 2)

      variations, activations = self.sample_completion_cloud(
          prompt=prompt,
          num_samples=num_samples,
          layer_idx=layer_idx,
          temperature=sample_temperature,
          top_p=sample_top_p,
          max_new_tokens=max_new_tokens,
      )

      print("\nActivation matrix shape:", activations.shape)

      max_components = min(n_axes, activations.shape[0], activations.shape[1])
      pca = PCA(n_components=max_components)
      scores = pca.fit_transform(activations)

      print("\nPCA explained variance ratios:")
      for i, ratio in enumerate(pca.explained_variance_ratio_):
          print(f"  PC{i + 1}: {ratio:.4f}")

      baseline_text = self.generate_text(
          prompt=prompt,
          system_prompt="You are a helpful assistant.",
          max_new_tokens=max_new_tokens,
          do_sample=True,
          temperature=0.4,
          top_p=0.9,
      )

      axes: List[ControlAxis] = []

      for i in range(max_components):
          pc = pca.components_[i].astype(np.float32)
          pc = pc / (np.linalg.norm(pc) + 1e-8)

          pc_std = float(np.sqrt(max(float(pca.explained_variance_[i]), 1e-12)))
          variance_ratio = float(pca.explained_variance_ratio_[i])

          component_scores = scores[:, i]
          pos_sample_idx = int(np.argmax(component_scores))
          neg_sample_idx = int(np.argmin(component_scores))

          positive_sample = variations[pos_sample_idx]
          negative_sample = variations[neg_sample_idx]

          steering_layers, layer_weights = self.build_steering_layers(
              center_layer=layer_idx,
              radius=steering_radius,
              decay=0.7,
          )

          axis = ControlAxis(
              index=i,
              label=f"PC{i + 1}",
              vector=pc,
              positive_example="",
              negative_example="",
              layer_idx=layer_idx,
              steering_layers=steering_layers,
              layer_weights=layer_weights,
              variance=variance_ratio,
              scale=pc_std,
              source="pca",
              positive_sample=positive_sample,
              negative_sample=negative_sample,
          )

          print(f"\nGenerating pole outputs for PC{i + 1} with coefficient ±{pole_coefficient}...")

          negative_pole = self.steer_generation(
              prompt=prompt,
              axis=axis,
              coefficient=-pole_coefficient,
              max_new_tokens=max_new_tokens,
              temperature=0.4,
              top_p=0.9,
          )

          positive_pole = self.steer_generation(
              prompt=prompt,
              axis=axis,
              coefficient=+pole_coefficient,
              max_new_tokens=max_new_tokens,
              temperature=0.4,
              top_p=0.9,
          )

          judgment = self.label_and_orient_axis(
              user_prompt=prompt,
              negative_text=negative_pole,
              baseline_text=baseline_text,
              positive_text=positive_pole,
          )

          axis.label = judgment["label"]
          axis.coherence = judgment["coherence"]
          axis.negative_example = negative_pole
          axis.positive_example = positive_pole

          # PCA sign is arbitrary. Orient so positive slider means "more label".
          if not judgment["positive_pole_has_more_label"]:
              axis.vector = (-axis.vector).astype(np.float32)

              axis.positive_example, axis.negative_example = (
                  axis.negative_example,
                  axis.positive_example,
              )

              axis.positive_sample, axis.negative_sample = (
                  axis.negative_sample,
                  axis.positive_sample,
              )

              print(f"PC{i + 1}: flipped orientation so + means more {axis.label}.")

          print(f"Axis {i}: {axis.label}")
          print(f"  variance ratio: {axis.variance:.4f}")
          print(f"  PCA std scale:  {axis.scale:.4f}")
          print(f"  coherence:      {axis.coherence:.1f}/10")
          print(f"  judge rationale: {judgment.get('rationale', '')}")

          axes.append(axis)

      return axes, variations, baseline_text

  # ----------------------------
  # Optional custom axis.
  # This is NOT the paper method. It exists only so your current React UI's
  # "Add custom axis" button does not break.
  # ----------------------------
  def extract_custom_axis(
      self,
      concept: str,
      original_prompt: str,
      layer_idx: Optional[int] = None,
      max_new_tokens: int = 180,
  ) -> ControlAxis:
      if layer_idx is None:
          layer_idx = self.target_layer

      print(f"\n[Optional custom axis] Creating targeted contrastive axis for: {concept}")

      pos_sys = f"You are a helpful assistant. Respond in a style that is extremely {concept}."
      neg_sys = f"You are a helpful assistant. Respond in a style that is the opposite of {concept}."

      pos_text, pos_vec = self.generate_with_activation(
          prompt=original_prompt,
          system_prompt=pos_sys,
          layer_idx=layer_idx,
          max_new_tokens=max_new_tokens,
          temperature=0.7,
          top_p=0.9,
      )

      neg_text, neg_vec = self.generate_with_activation(
          prompt=original_prompt,
          system_prompt=neg_sys,
          layer_idx=layer_idx,
          max_new_tokens=max_new_tokens,
          temperature=0.7,
          top_p=0.9,
      )

      raw = pos_vec - neg_vec
      vec = raw.astype(np.float32)
      vec = vec / (np.linalg.norm(vec) + 1e-8)

      steering_layers, layer_weights = self.build_steering_layers(
          center_layer=layer_idx,
          radius=0,
          decay=0.7,
      )

      return ControlAxis(
          index=-1,
          label=self._clean_label(concept),
          vector=vec,
          positive_example=pos_text,
          negative_example=neg_text,
          layer_idx=layer_idx,
          steering_layers=steering_layers,
          layer_weights=layer_weights,
          variance=0.0,
          scale=1.0,
          source="custom_contrastive",
          coherence=0.0,
      )

  def axis_to_dict(self, axis: ControlAxis) -> Dict[str, Any]:
      return {
          "index": axis.index,
          "label": axis.label,
          "positive_example": axis.positive_example,
          "negative_example": axis.negative_example,
          "variance": axis.variance,
          "coherence": axis.coherence,
          "source": axis.source,
      }


print("PCAControlSystem class loaded.")
```

```python
from fastapi import FastAPI
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
import uvicorn
import nest_asyncio

app = FastAPI(title="PCA Latent-Steer API")

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

# Global server state
global_axes: List[ControlAxis] = []
global_variations: List[str] = []
global_baseline: str = ""
global_last_prompt: str = ""
generation_cache: Dict[Tuple[str, int, float], str] = {}

# Paper-method defaults
DEFAULT_NUM_SAMPLES = 30
DEFAULT_N_AXES = 2
DEFAULT_POLE_COEFFICIENT = 2.0

# Your React slider sends -50..50.
# Internally we want PCA coefficient units roughly -3..3.
UI_SLIDER_MAX = 50.0
PCA_ALPHA_MAX = 3.0


def ui_coefficient_to_alpha(coefficient: float) -> float:
    """
    If coefficient looks like a UI slider value, map -50..50 to -3..3.
    If coefficient already looks like a PCA alpha, leave it alone.

    Examples:
      UI +50 -> alpha +3
      UI +10 -> alpha +0.6
      direct API +2 -> alpha +2
    """

    coefficient = float(coefficient)

    # Direct API mode: user sent an already-normalized alpha.
    if abs(coefficient) <= PCA_ALPHA_MAX:
        return coefficient

    # UI slider mode.
    alpha = coefficient / UI_SLIDER_MAX * PCA_ALPHA_MAX
    alpha = max(-PCA_ALPHA_MAX, min(PCA_ALPHA_MAX, alpha))
    return float(alpha)


class DiscoverRequest(BaseModel):
    prompt: str
    num_variations: int = DEFAULT_NUM_SAMPLES
    n_axes: int = DEFAULT_N_AXES


class SteerRequest(BaseModel):
    prompt: str
    axis_index: int
    coefficient: float
    max_tokens: int = 180


class CustomAxisRequest(BaseModel):
    concept: str


@app.get("/")
async def root():
    return {
        "status": "ok",
        "message": "PCA Latent-Steer API is running.",
        "docs": "http://127.0.0.1:8000/docs",
    }


@app.post("/discover")
async def discover_endpoint(req: DiscoverRequest):
    global global_axes
    global_variations
    global_baseline
    global_last_prompt
    global generation_cache

    print(f"\nAPI /discover prompt: {req.prompt}")

    generation_cache.clear()
    global_last_prompt = req.prompt

    # Force paper method defaults unless caller asks for more samples.
    num_samples = max(int(req.num_variations), DEFAULT_NUM_SAMPLES)
    n_axes = min(int(req.n_axes), DEFAULT_N_AXES)

    axes, variations, baseline = system.discover_pca_axes(
        prompt=req.prompt,
        num_samples=num_samples,
        n_axes=n_axes,
        pole_coefficient=DEFAULT_POLE_COEFFICIENT,
        sample_temperature=1.15,
        steering_radius=0,
        max_new_tokens=180,
    )

    global_axes = axes
    global_variations = variations
    global_baseline = baseline

    # Cache baseline for every axis at alpha 0.
    for ax in global_axes:
        generation_cache[(req.prompt, ax.index, 0.0)] = baseline

    return {
        "status": "success",
        "axes": [system.axis_to_dict(ax) for ax in global_axes],
        "variations": global_variations,
        "baseline": global_baseline,
    }


@app.post("/steer")
async def steer_endpoint(req: SteerRequest):
    global generation_cache

    if not global_axes:
        return {
            "status": "error",
            "message": "No axes available. Run /discover first.",
        }

    if req.axis_index < 0 or req.axis_index >= len(global_axes):
        return {
            "status": "error",
            "message": f"Invalid axis_index {req.axis_index}.",
        }

    axis = global_axes[req.axis_index]
    alpha = ui_coefficient_to_alpha(req.coefficient)
    alpha_key = round(alpha, 4)

    cache_key = (req.prompt, req.axis_index, alpha_key)

    if cache_key in generation_cache:
        print(f"API /steer cache hit: axis={axis.label}, alpha={alpha_key}")
        return {
            "status": "success",
            "generated_text": generation_cache[cache_key],
            "alpha": alpha_key,
        }

    print(f"\nAPI /steer axis={axis.label}, UI coefficient={req.coefficient}, alpha={alpha_key}")

    generated = system.steer_generation(
        prompt=req.prompt,
        axis=axis,
        coefficient=alpha,
        max_new_tokens=req.max_tokens,
        temperature=0.4,
        top_p=0.9,
    )

    generation_cache[cache_key] = generated

    return {
        "status": "success",
        "generated_text": generated,
        "alpha": alpha_key,
    }


@app.post("/custom_axis")
async def custom_axis_endpoint(req: CustomAxisRequest):
    """
    Optional compatibility endpoint for your existing React UI.

    Note:
    This is NOT the paper PCA method. It creates a targeted contrastive axis
    from the user-provided concept. It is included only so the current UI's
    custom-axis button remains usable.
    """

    global global_axes
    global global_last_prompt

    if not global_last_prompt:
        return {
            "status": "error",
            "message": "Run /discover first so the server knows the original prompt.",
        }

    custom_axis = system.extract_custom_axis(
        concept=req.concept,
        original_prompt=global_last_prompt,
        max_new_tokens=180,
    )

    custom_axis.index = len(global_axes)
    global_axes.append(custom_axis)

    return {
        "status": "success",
        "axis": system.axis_to_dict(custom_axis),
    }


nest_asyncio.apply()

config = uvicorn.Config(
    app,
    host="127.0.0.1",
    port=8000,
    log_level="info",
)

server = uvicorn.Server(config)

server_thread = threading.Thread(
    target=server.run,
    daemon=True,
)

server_thread.start()

print("\nLocal PCA Latent-Steer API running at:")
print("  http://127.0.0.1:8000")
print("Docs:")
print("  http://127.0.0.1:8000/docs")
```
